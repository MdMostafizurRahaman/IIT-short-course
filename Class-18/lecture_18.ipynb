````python
import numpy as np
from sklearn.linear_model import LinearRegression

# Data: Study hours vs Marks
X = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)
y = np.array([50, 65, 75, 85, 95])

# Create linear regression model
model = LinearRegression()
model.fit(X, y)

print(model.coef_)
print(model.intercept_)

# Predict marks for a new student (e.g., 5 hours)
new_hours = np.array([[5]])
predicted_marks = model.predict(new_hours)
print("Predicted marks for 5 study hours:", predicted_marks)

# Prediction for multiple values at once (Optional)
new_hours = np.array([[3], [7], [8]])
predicted_marks = model.predict(new_hours)
print("Predicted marks for multiple study hours:", predicted_marks)
````

````python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# df = pd.read_csv("house_prices_linear.csv")

# Step 1: Define Features and Target
y = df["Price"]
X = df[["Size", "Bedrooms", "Age"]]

# Step 2: Split Data into Training and Testing Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Step 3: Train the Linear Regression Model
regressor = LinearRegression() # Initialize model
regressor.fit(X_train, y_train) # Train model

# Step 4: Examine the Regression Equation
print("Intercept:", regressor.intercept_)
print("Coefficients:", regressor.coef_)

# Step 5: Make Predictions
y_pred = regressor.predict(X_test) # Predicted prices

# Step 6: Evaluate Model Performance
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)

r2 = r2_score(y_test, y_pred)
print("R^2 Score:", r2)
````

## Important Note 

. Intercept: The predicted value when all features are zero.

. Coefficients: The contribution of each feature to the predicted price.

. MSE: Measures the average squared difference between predicted and actual prices; smaller is
better.

. R2 Score: Proportion of variance in the target variable explained by the model; closer to 1 indicates
better fit.

. R-squared (R2): Measures how well the regression line explains the variance in the data. Better if
high (closer to 1).

. Mean Squared Error (MSE): Average squared difference between predicted and actual values.
Better if low (closer to 0).

· Mean Absolute Error (MAE): Average absolute difference between predicted and actual values.
Better if low.

### 1. **Subplotting** with Matplotlib

Create **multiple plots** in one figure for easy comparison.

**Two methods:**
1. `plt.subplots()` - **Recommended** for complex layouts
2. `plt.subplot()` - **Simple** for basic cases

**Demo:** Student exam scores comparison

````python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
````

````python
# Set style for matplotlib
plt.style.use('default')

# For reproducibility (numpy)
np.random.seed(42)
````

````python
import random

# Method 1: Using plt.subplots() - Recommended for complex layouts
print("Method 1: Using plt.subplots()")

# Student data with realistic names and slightly randomized scores (X)
students = ['Ahmed', 'Fatima', 'Omar', 'Aisha', 'Mohammed', 'Sara', 'Ali', 'Noor', 'Hassan', 'Zara']

# Base scores with some randomization for realism (Y)
exam1 = [88, 76, 91, 65, 82, 74, 95, 69, 87, 78]
exam2 = [79, 59, 90, 72, 85, 77, 85, 68, 80, 81]
exam3 = [85, 73, 88, 66, 83, 79, 96, 66, 89, 75]

average = [sum([e1,e2,e3])/3 for e1, e2, e3 in zip(exam1, exam2, exam3)] # (Y)
````

````python
# Create a 2x2 subplot grid using plt.subplots()
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Plot Exam 1 scores
ax[0][0].bar(students, exam1, color='skyblue', alpha=0.7)
ax[0][0].set_title('Exam 1 Scores')
ax[0][0].set_xlabel('Student Name')
ax[0][0].set_ylabel('Score')
ax[0][0].set_ylim(40, 100)
ax[0][0].tick_params(axis='x', rotation=45)
ax[0][0].grid(True, alpha=0.3)

# Plot Exam 2 scores
ax[0][1].bar(students, exam2, color='lightgreen', alpha=0.7)
ax[0][1].set_title('Exam 2 Scores')
ax[0][1].set_xlabel('Student Name')
ax[0][1].set_ylabel('Score')
ax[0][1].set_ylim(40, 100)
ax[0][1].tick_params(axis='x', rotation=45)
ax[0][1].grid(True, alpha=0.3)

# Plot Exam 3 scores
ax[1][0].bar(students, exam3, color='lightcoral', alpha=0.7)
ax[1][0].set_title('Exam 3 Scores')
ax[1][0].set_xlabel('Student Name')
ax[1][0].set_ylabel('Score')
ax[1][0].set_ylim(40, 100)
ax[1][0].tick_params(axis='x', rotation=45)
ax[1][0].grid(True, alpha=0.3)

# Plot Average scores
ax[1][1].bar(students, average, color='gold', alpha=0.7)
ax[1][1].set_title('Average Scores')
ax[1][1].set_xlabel('Student Name')
ax[1][1].set_ylabel('Score')
ax[1][1].set_ylim(40, 100)
ax[1][1].tick_params(axis='x', rotation=45)
ax[1][1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
print("Method 2: Using plt.subplot() - Good for simple cases")

# Method 2: Using plt.subplot()
plt.figure(figsize=(12, 8))

# Subplot 1: Exam 1
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, position 1

plt.bar(students, exam1, color='skyblue', alpha=0.7)

plt.title('Exam 1 Scores')
plt.xlabel('Student Name')
plt.ylabel('Score')

plt.ylim(40, 100)

plt.xticks(rotation=45)

plt.grid(True, alpha=0.3)

# Subplot 2: Exam 2
plt.subplot(2, 2, 2)  # 2 rows, 2 columns, position 2
plt.bar(students, exam2, color='lightgreen', alpha=0.7)
plt.title('Exam 2 Scores')
plt.xlabel('Student Name')
plt.ylabel('Score')
plt.ylim(40, 100)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Subplot 3: Exam 3
plt.subplot(2, 2, 3)  # 2 rows, 2 columns, position 3
plt.bar(students, exam3, color='lightcoral', alpha=0.7)
plt.title('Exam 3 Scores')
plt.xlabel('Student Name')
plt.ylabel('Score')
plt.ylim(40, 100)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Subplot 4: Average
plt.subplot(2, 2, 4)  # 2 rows, 2 columns, position 4
plt.bar(students, average, color='gold', alpha=0.7)
plt.title('Average Scores')
plt.xlabel('Student Name')
plt.ylabel('Score')
plt.ylim(40, 100)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nBoth methods create the same result, but plt.subplots() is more flexible for complex layouts.")
````

**Comparison:** Matplotlib vs Seaborn approach

````python
import seaborn as sns
````

````python
# Simple comparison: Matplotlib vs Seaborn
print("Let's compare creating the same bar chart with Matplotlib and Seaborn:")

# First, create some sample data for demonstration
sample_data = pd.DataFrame({
    'class': ['A', 'B', 'C'],
    'average_score': [85, 78, 72]
})
````

````python
print("\nMatplotlib approach (more code):")

plt.figure(figsize=(8, 5))

# Matplotlib version
bars = plt.bar(sample_data['class'], sample_data['average_score'], alpha=0.7)

plt.title('Average Scores by Class (Matplotlib)')
plt.xlabel('Class')
plt.ylabel('Average Score')
plt.ylim(65, 90)

# show values over the bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'{height:.1f}', ha='center', va='bottom')


plt.grid(True, alpha=0.3, axis='y')
plt.show()
````

````python
print("\nSeaborn approach (less code, better defaults):")

plt.figure(figsize=(8, 5))

# Seaborn version
sns.barplot(data=sample_data, x='class', y='average_score', hue='average_score', legend=False)

plt.title('Average Scores by Class (Seaborn)')
plt.xlabel('Class')
plt.ylabel('Average Score')
plt.ylim(65, 90)
plt.grid(True, alpha=0.3, axis='y')
plt.show()

print("Notice how Seaborn:")
print("- Uses better default colors")
print("- Automatically handles the data format")
print("- Creates cleaner, more professional-looking plots")
print("- Requires less code for the same result")
````

### 4. **Package Management** & Versions

#### **Version Issues**
**Breaking changes** in updates can break your code:
- Function names change
- API modifications  
- Dependency conflicts

**Example:** Code works with `pandas 1.5.0`, breaks with `pandas 2.0.0`

#### **Why Track Versions?**
- **Reproducibility** - Exact same setup for everyone
- **Stability** - Avoid unexpected breaks
- **Collaboration** - Team uses identical versions
- **Debugging** - Know exactly what caused issues

#### **requirements.txt**
**Simple file** listing exact package versions for your project.

**Benefits:**
- **Easy sharing** - One file installs everything
- **Version control** - Track with Git
- **Automation** - CI/CD can recreate environment
- **Documentation** - Clear dependency record

**Usage:**
```bash
pip freeze > requirements.txt  # Create
pip install -r requirements.txt  # Install
```

**Example:**
```
numpy==1.24.3
pandas==2.0.3
matplotlib==3.7.2
```

**Prevents:** "Works on my machine" issues!

### 5. **Python Environments** Made Simple

#### **Why Environments?**
**Project isolation** prevents package conflicts.

**Example:** Project A needs `pandas 1.5.0`, Project B needs `pandas 1.5.2`

**Benefits:**
- **No conflicts** - Each project has own packages
- **Reproducible** - Exact setup recreation
- **Clean system** - Main Python stays tidy

#### **Quick Start with venv**
Python's built-in tool (free, no extra install).

**Steps:**
1. **Open Terminal** (Command Prompt/PowerShell on Windows)
2. **Navigate:** `cd your-project-folder`
3. **Create:** `python -m venv myenv`
4. **Activate:**
   - **Windows:** `myenv\Scripts\activate`
   - **Linux/Mac:** `source myenv/bin/activate`
5. **Install:** `pip install package-name`
6. **Work** on your project
7. **Deactivate:** `deactivate`

**Tips:**
- See `(myenv)` in terminal when active
- Use descriptive names: `data_project`
- Delete folder to remove environment

#### **Conda** (Data Science Alternative)
**Advanced package manager** for scientific computing.

**Setup:** Download Miniconda → https://www.anaconda.com/docs/getting-started/miniconda/install

**Why Conda?**
- Handles **complex packages** (CUDA, MKL, etc.)
- **Cross-platform** compatibility
- **Pre-compiled** scientific packages

**Commands:**
```bash
conda create -n myenv python=3.9
conda activate myenv
conda install numpy pandas matplotlib
```

**Use venv** for simple projects, **Conda** for data science.

**Check current environment:**

````python
# Check current Python environment
import sys
import os

print("🔍 Current Python Environment Information:")
print("=" * 50)
print(f"Python version: {sys.version}")
print(f"Python executable: {sys.executable}")
print(f"Current working directory: {os.getcwd()}")

# Check if we're in a virtual environment
if hasattr(sys, 'real_prefix') or (hasattr(sys, 'base_prefix') and sys.base_prefix != sys.prefix):
    print("✅ We are in a virtual environment!")
    print(f"Environment path: {sys.prefix}")
else:
    print("ℹ️  We are using the system Python (not in a virtual environment)")

print("\n📦 Installed Data Science Packages:")
print("-" * 40)

# Check core packages with better error handling
def check_package(package_name, import_name=None):
    if import_name is None:
        import_name = package_name
    try:
        module = __import__(import_name)
        version = getattr(module, '__version__', 'unknown')
        return f"✅ {package_name}: {version}"
    except ImportError:
        return f"❌ {package_name}: Not installed"
    except Exception as e:
        return f"⚠️  {package_name}: Error checking version ({str(e)})"

packages_to_check = [
    ('numpy', 'numpy'),
    ('pandas', 'pandas'),
    ('matplotlib', 'matplotlib'),
    ('seaborn', 'seaborn'),
    ('plotly', 'plotly'),
    ('scikit-learn', 'sklearn'),
    ('scipy', 'scipy')
]

for package_name, import_name in packages_to_check:
    print(check_package(package_name, import_name))

print("\n💡 Tips:")
print("- conda env list → see all environments")
print("- pip freeze > requirements.txt → save packages")
print("- Always activate environment before working!")
print("\n🔄 Recreate: pip install numpy pandas matplotlib seaborn plotly scikit-learn")
````
