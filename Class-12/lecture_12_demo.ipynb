Use `head`, `tail`, `info`, and `describe` to inspect the DataFrame.

```python
first_five_rows = df.head()
print(f">> First 5 rows:\n{first_five_rows}")

last_five_rows = df.tail()
print(f"\n\n>> Last 5 rows:\n{last_five_rows}")

print("\n\n>> DataFrame info:")
df.info()

print(f"\n\n>> Statistical summary:\n{df.describe()}")
```

## Handle Data Types

Check and convert data types, such as converting hire_date to datetime.

```python
column_data_types = df.dtypes
print(f"Data types of all columns:\n{column_data_types}")

salary_dtype = df['salary'].dtype
hire_date_original_dtype = df['hire_date'].dtype
print(f"\n\n>> Salary data type: {salary_dtype}, Hire date original type: {hire_date_original_dtype}")

df['hire_date'] = pd.to_datetime(df['hire_date'])
hire_date_converted_dtype = df['hire_date'].dtype
print(f"\n\n>> Hire date after conversion: {hire_date_converted_dtype}")
```

## Compute Statistical Summaries

Calculate mean, median, standard deviation, min, max, value counts, and sums for columns.

```python
mean_salary = df['salary'].mean()
median_salary = df['salary'].median()
std_years_experience = df['years_experience'].std()
min_salary = df['salary'].min()
max_salary = df['salary'].max()
department_counts = df['department'].value_counts()
total_salaries = df['salary'].sum()

print(f"Mean salary: {mean_salary}")
print(f"Median salary: {median_salary}")
print(f"Std Dev of years_experience: {std_years_experience}")
print(f"Min salary: {min_salary}, Max salary: {max_salary}")
print(f"\nDepartment counts:\n{department_counts}")
print(f"\nTotal salaries: {total_salaries}")
```

## Check for Missing Values

Identify missing values per column and total missing values in the DataFrame.

```python
missing_values_per_column = df.isnull().sum()
total_missing_values = df.isnull().sum().sum()
print(f"Missing values per column:\n{missing_values_per_column}")
print(f"Total missing values in DataFrame: {total_missing_values}")
```

## Select Columns

Select single columns, multiple columns, and columns by position.

```python
single_name_column = df['name'].head()
print(f"Single column (name):\n{single_name_column}")

multiple_cols_selected = df[['name', 'department', 'salary']].head()
print(f"\n\n>> Multiple columns (name, department, salary):\n{multiple_cols_selected}")

columns_by_position = df.iloc[:, [0, 2, 5]].head()
print(f"\n\n>> Columns by position (0, 2, 5):\n{columns_by_position}")
```

## Filter Data

Filter rows based on conditions like salary thresholds, department, experience, and certifications.

```python
high_salary_employees = df[df['salary'] > 80000][['name', 'salary']]
print(f"Employees with salary > 80000:\n{high_salary_employees}")

it_department_employees = df[df['department'] == 'IT'][['name', 'department', 'position']]
print(f"\n\n>> IT department employees:\n{it_department_employees}")

experienced_employees = df[df['years_experience'] >= 5][['name', 'years_experience']]
print(f"\n\n>> Employees with years_experience >= 5:\n{experienced_employees}")

certified_employees = df[df['has_certification'] == True][['name', 'has_certification']]
print(f"\n\n>> Employees with certification:\n{certified_employees}")

it_high_salary = df[(df['department'] == 'IT') & (df['salary'] > 90000)][['name', 'department', 'salary']]
print(f"\n\n>> IT employees with salary > 90000:\n{it_high_salary}")
```

## Create New Columns

Create new columns using arithmetic operations, apply functions, and string concatenation.

```python
df['annual_bonus'] = df['salary'] * 0.10
bonus_data = df[['name', 'salary', 'annual_bonus']].head()
print(f"New column 'annual_bonus' (10% of salary):\n{bonus_data}")

df['role_description'] = df['name'] + ' - ' + df['position']
role_desc_data = df[['role_description', 'salary']].head()
print(f"\n\n>> New column 'role_description':\n{role_desc_data}")
```

## Sort Data

Sort the DataFrame by single columns, multiple columns, ascending or descending.

```python
sorted_by_salary_desc = df.sort_values('salary', ascending=False)[['name', 'salary']].head()
print(f"Sort by salary (descending):\n{sorted_by_salary_desc}")

sorted_by_experience = df.sort_values('years_experience', ascending=True)[['name', 'years_experience']].head()
print(f"\n\n>> Sort by years_experience (ascending):\n{sorted_by_experience}")

sorted_multi_column = df.sort_values(['salary','name'], ascending=[True, False])[['name', 'department', 'salary']].head()
print(f"\n\n>> Sort by salary then name:\n{sorted_multi_column}")
```

## Group and Aggregate Data

Group by columns and perform aggregations like mean, size, and multiple stats.

```python
avg_salary_by_dept = df.groupby('department')['salary'].mean()
print(f"Average salary by department:\n{avg_salary_by_dept}")

employee_count_by_dept = df.groupby('department').size()
print(f"\n\n>> Number of employees by department:\n{employee_count_by_dept}")

dept_salary_stats = df.groupby('department')['salary'].agg(['mean', 'min', 'max', 'count'])
print(f"\n\n>> Salary statistics by department:\n{dept_salary_stats}")
```

## Save Data to CSV

Save the full DataFrame, filtered data, and selected columns to CSV files.

```python
df.to_csv('output_full.csv', index=False)
print("Full dataset saved to 'output_full.csv'")

it_employees_data = df[df['department'] == 'IT']
it_employees_data.to_csv('output_it_employees.csv', index=False)
print("IT employees saved to 'output_it_employees.csv'")

name_salary_data = df[['name', 'salary']]
name_salary_data.to_csv('output_name_salary.csv', index=False)
print("Name and salary saved to 'output_name_salary.csv'")
```

## Common Patterns and Queries

Find unique values, number of uniques, oldest and newest employees, and department-specific data.

```python
unique_departments = df['department'].unique()
num_unique_departments = df['department'].nunique()
print(f"Unique departments: {unique_departments}")
print(f"Number of unique departments: {num_unique_departments}")

oldest_employee = df.loc[df['hire_date'].idxmin()]
print(f"\n\n>> Oldest employee: {oldest_employee['name']}, Hire Date: {oldest_employee['hire_date']}, Department: {oldest_employee['department']}")

newest_employee = df.loc[df['hire_date'].idxmax()]
print(f"\n\n>> Newest employee: {newest_employee['name']}, Hire Date: {newest_employee['hire_date']}, Department: {newest_employee['department']}")

research_employees = df[df['department'] == 'Research'][['name', 'salary']]
print(f"\n\n>> Research department employees:\n{research_employees}")
```
