## Lecture 14: Column Operations With Pandas

Dataset: `person_info.csv`

### Import Libraries

```python
import pandas as pd
```

### Download Data

```python
df = pd.read_csv('https://raw.githubusercontent.com/arman18/data/refs/heads/main/person_info.csv')
df.to_csv('person_info.csv', index=False)
```

### 1. Data Exploration

Before performing any operations, it's important to explore the dataset to understand its structure, data types, and any potential issues like missing values or incorrect data types.

Let's start by loading the data and exploring it.

```python
# Load the dataset
df = pd.read_csv('person_info.csv')

# Display the first 5 rows to see the data
print("First 5 rows of the dataset:")
df.head()
```

```python
# Get information about the dataset: columns, data types, non-null counts
print("\nDataset information:")
df.info()
```

```python
# Get summary statistics for numerical columns
print("\nSummary statistics:")
df.describe()
```

### 2. Column Subsetting

Column subsetting means selecting specific columns from the DataFrame. This is useful when you only need certain columns for analysis.

You can select columns by name **using square brackets** or the `.loc` accessor.

```python
# Select specific columns by name
selected_columns = df[['ID', 'Age', 'Gender', 'Income']]
print("Selected columns (ID, Age, Gender, Income):")
selected_columns.head()
```

```python
# Select columns using .loc (label-based selection)
# .loc[row_indexer, column_indexer]
height_weight = df.loc[:, ['Height', 'Weight']]
print("\nHeight and Weight columns:")
height_weight.head()
```

### 3. DataFrame Filtering

Filtering allows you to select rows that meet certain conditions. This is useful for focusing on specific subsets of data.

Use boolean indexing with comparison operators.

```python
# Filter rows where Age is greater than 30
older_people = df[df['Age'] > 30]
print("People older than 30:")
older_people[['ID', 'Age', 'Gender']].head()

# Filter rows where Gender is 'F' (Female)
females = df[df['Gender'] == 'F']
print("\nFemale participants:")
females[['ID', 'Age', 'Gender', 'Income']].head()

# Combine conditions: Age > 25 AND Income > 50000
young_high_income = df[(df['Age'] > 25) & (df['Income'] > 50000)]
print("\nPeople older than 25 with income > 50000:")
young_high_income[['ID', 'Age', 'Income']].head()
```

### 4. Column Renaming

Sometimes column names need to be changed for clarity or consistency. You can rename columns using the .rename() method.

```python
# Rename specific columns
df_renamed = df.rename(columns={
    'Height': 'height_cm',
    'Weight': 'weight_kg',
    'Income': 'annual_income',
    'Health Condition': 'health_status'
})

print("Columns after renaming:")
print(df_renamed.columns.tolist())

# Display a sample
df_renamed[['ID', 'height_cm', 'weight_kg', 'annual_income']].head()
```

### 5. Column Name Standardizing

To make column names consistent, we can convert them to lowercase and replace spaces with underscores. This is a common practice in data analysis.

Use string methods on the columns attribute.

```python
# Standardize column names: convert to lowercase and replace spaces with underscores
cols = df.columns 

# covert the column names to lower case
cols = [col.lower() for col in cols]

# replace spaces with underscores
cols = [col.replace(' ', '_') for col in cols]

df.columns = cols

print("Standardized column names:")
print(df.columns.tolist())

# Display the first few rows with new column names
df.head()
```

### 6. Safe Data Type Conversion

Sometimes columns have the wrong data type or contain invalid values. Safe conversion uses `coerce` **to convert invalid values to `NaN` instead of raising errors**.

Use `pd.to_numeric()` for numbers, `pd.to_datetime()` for dates, etc.

```python
# Convert height to numeric (it contains strings like '(160)', 'Tall')
df['height'] = pd.to_numeric(df['height'], errors='coerce')
```

```python
# Convert weight to numeric (it contains negative values and large numbers)
df['weight'] = pd.to_numeric(df['weight'], errors='coerce')
```

```python
# Convert income to numeric (should already be, but safe)
df['income'] = pd.to_numeric(df['income'], errors='coerce')
```

```python
print("Data types after conversion:")
print(df.dtypes)
```

```python
# Check for NaN values created by coercion
print("\nNumber of NaN values in each column:")
print(df.isnull().sum())
```

### 7. Column Modification

You can modify existing columns or create new ones using mathematical operations or string manipulations.

This includes calculations, transformations, and cleaning categorical data.

```python
# Mathematical calculation: Create BMI column
# BMI = weight (kg) / (height (m))^2
df['bmi'] = df['weight'] / (df['height'] / 100) ** 2

print("Added BMI column:")
df[['id', 'height', 'weight', 'bmi']].head()
```

```python
# Modify categorical data: Standardize gender values
df['gender'] = df['gender'].str.upper()

# Create age groups
df['age_group'] = pd.cut(df['age'], bins=[0, 25, 35, 50, 100], labels=['Young', 'Adult', 'Middle-aged', 'Senior'])
```

```python

print("\nAdded age_group column:")
df[['id', 'age', 'gender', 'age_group']].head()
```

```python
# Clean employment status (capitalize first letter)
df['employment'] = df['employment'].str.capitalize()

print("\nCleaned employment column:")
df[['id', 'employment']].head()
```
