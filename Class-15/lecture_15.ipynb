### 2. Creating New Columns

Creating new columns in pandas is straightforward. You can create columns by assigning values directly, using mathematical operations, or applying functions to existing columns.

```python
# Create a new column by performing calculations on existing columns
# First, ensure numeric types
df['Height'] = pd.to_numeric(df['Height'], errors='coerce')
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')

# BMI calculation: weight (kg) / (height (m))^2
df['BMI'] = df['Weight'] / (df['Height'] / 100) ** 2

print("Added BMI column:")
df[['ID', 'Height', 'Weight', 'BMI']].head()
```

### 3. Creating Columns with Dummy Values

Sometimes you need to create columns with constant values or placeholders. This can be useful for adding flags, categories, or temporary data.

```python
# Create a column with a constant value
df['Survey_Year'] = 2024

# Create a column with a string placeholder
df['Status'] = 'Active'

# Create a column with boolean values
df['Is_Processed'] = False

print("Added dummy columns:")
df[['ID', 'Survey_Year', 'Status', 'Is_Processed']].head()
```

### 4. Creating Columns with apply() Method

The `apply()` method allows you to apply a function along an axis of the DataFrame. There are two main ways to use it:

- **Series.apply()**: Apply a function to each element in a Series (column)
- **DataFrame.apply()**: Apply a function along rows or columns of the DataFrame

This is useful for more complex transformations that can't be done with simple arithmetic operations.

```python
df_demo = pd.DataFrame({
    'Name': ["Abul", "Babul", "Cabul"],
    'Age': [20, 30, 40],
    'Salary': [50000, 60000, 70000] 
})

def row_summary(row):
    return f"{row['Name']} is {row['Age']} years old"

df_demo['Summary'] = df_demo.apply(row_summary, axis=1)
print(df_demo)
```

```python
# Series.apply(): Apply a function to each element in a column
def categorize_age(age):
    if age < 25:
        return 'Young'
    elif age < 35:
        return 'Adult'
    elif age < 50:
        return 'Middle-aged'
    else:
        return 'Senior'

df['Age_Category'] = df['Age'].apply(categorize_age)

print("Added Age_Category using Series.apply():")
df[['ID', 'Age', 'Age_Category']].head()
```

```python
# DataFrame.apply(): Apply a function along rows (axis=1) or columns (axis=0)
def calculate_health_score(row):
    # Simple health score based on BMI and age
    bmi_score = 10 if row['BMI'] < 25 else (5 if row['BMI'] < 30 else 0)
    age_score = 10 if row['Age'] < 40 else (5 if row['Age'] < 60 else 0)
    return bmi_score + age_score

df['Health_Score'] = df.apply(calculate_health_score, axis=1)

print("Added Health_Score using DataFrame.apply():")
df[['ID', 'Age', 'BMI', 'Health_Score']].head()
```

### 5. Creating Columns with Lambda Functions

Lambda functions are small, anonymous functions that can be used inline. They're particularly useful with `apply()` for simple transformations. The syntax is: `lambda arguments: expression`

```python
# Using lambda with Series.apply() for simple transformations
df['Income_Category'] = df['Income'].apply(lambda x: 'High' if x > 60000 else ('Medium' if x > 30000 else 'Low'))

# Convert height from cm to inches using lambda
df['Height_Inches'] = df['Height'].apply(lambda x: x * 0.3937 if pd.notnull(x) else None)

# Create a full name column (assuming we have first and last names, but using ID for demo)
df['Person_Code'] = df['ID'].apply(lambda x: f'Person_{x}')

print("Added columns using lambda functions:")
df[['ID', 'Income', 'Income_Category', 'Height', 'Height_Inches', 'Person_Code']].head()
```

### 6. Sorting Values by Single and Multiple Columns

Sorting data is essential for analysis. You can sort by one column or multiple columns using the `sort_values()` method. Use `ascending=False` for descending order.

```python
# Sort by a single column (ascending by default)
sorted_by_age = df.sort_values('Age')
print("Sorted by Age (ascending):")
sorted_by_age[['ID', 'Age', 'Gender']].head()
```

```python
# Sort by a single column in descending order
sorted_by_income_desc = df.sort_values('Income', ascending=False)
print("\nSorted by Income (descending):")
sorted_by_income_desc[['ID', 'Income']].head()
```

```python
# Sort by multiple columns: first by Gender, then by Age within each gender
sorted_multi = df.sort_values(['Gender', 'Age'])
print("\nSorted by Gender then Age:")
sorted_multi[['ID', 'Gender', 'Age']].head(10)
```

### 7. Deleting Columns

You can remove columns that are no longer needed using the `drop()` method. Use `axis=1` to specify columns (axis=0 for rows). The `inplace=True` parameter modifies the DataFrame directly.

```python
# Delete a single column
df_dropped = df.drop('Survey_Year', axis=1)
print("Columns after dropping 'Survey_Year':")
print(df_dropped.columns.tolist()[:10])  # Show first 10 columns
```

```python
# Delete multiple columns
columns_to_drop = ['Status', 'Is_Processed', 'Person_Code']
df_dropped_multi = df.drop(columns_to_drop, axis=1)
print("\nColumns after dropping multiple columns:")
print(df_dropped_multi.columns.tolist())
```

```python
# Delete columns in-place (modifies the original DataFrame)
df.drop(['Height_Inches'], axis=1, inplace=True)
print("\nOriginal DataFrame columns after in-place deletion:")
print(df.columns.tolist())
```
